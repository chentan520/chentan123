```python
## 1:
import random
import math
n = random.randint(1, 100)  # 函数 random.randint(a,b) 返回从 a 到 b 之间随机选出的一个整数
a, b, cnt = 1, 100, 0
while a <= b:
    cnt = cnt + 1
    m = math.ceil((a + b) / 2)  # 函数 math.ceil(x) 的返回值为数值 x 的向上取整
    if n < m:
        b = m - 1
    elif n > m:
        a = m + 1
    else:
        break
print(cnt)

```

```python
## 3:
def Change(radix, num):
    g = []
    while num != 0:
        r = num % radix
        num = num // radix
        g.append(r)
    ans = ''
    t = ('A', 'B', 'C', 'D', 'E', 'F')
    for x in range(len(g)):
        if g[x] >= 10:
            ans +=  t[g[x] - 10]
        else:
            ans += str(g[x])
    return ans[::-1]
print(Change(16, 12455))

```

```python
##  数据分析
# 1️⃣ 计算司机的平均评分（Average Driver Rating）
# ●从 drivers.csv 文件中计算 rating 列的平均值。
# ●将结果保存为：
# ○insight_type: "average_driver_rating"
# ○value: 计算出的平均评分

# 2️⃣ 计算拥有第二语言的司机比例（Percentage of Drivers with a Second Language）
# ●计算拥有第二语言的司机的比例（即 second_language 不为 "no" 的司机占总司机数的百分比）。
# ●将结果保存为：
# ○insight_type: "percentage_drivers_with_second_language"
# ○value: 计算出的百分比

# 3️⃣ 计算乘车成功率（Ride Success Rate）
# ●将所有 rides_{i}.csv 文件（从 1 到 4）合并为一个整体数据集。
# ●计算成功乘车（即 status == "Success"）的比例。
# ●将结果保存为：
# ○insight_type: "ride_success_rate"
# ○value: 计算出的百分比

# 输出要求（Output Requirements）
# ●将结果保存到一个名为 analysis_results.csv 的 CSV 文件中 （用代码保存）。
# ●该 CSV 文件应包含两列：
# ○insight_type
# ○value
# ●每一行对应一个任务结果，insight_type 为任务名称，value 为计算结果。
# ●所有数值在保留两位小数的范围内视为正确。
import pandas as pd


tdriver_file = r'D:\chentan\RNN_Numpy\driver.csv'
kdriver_data = pd.read_csv(tdriver_file)
avg_rating = kdriver_data['rating'].mean()
first_analysis_result = {
    'insight_type': 'average_driver_rating',
    'value': round(avg_rating, 2)
}

drivers_with_second_lang = kdriver_data[kdriver_data['second_language'] != 'no']
second_lang_percentage = (len(drivers_with_second_lang) / len(kdriver_data)) * 100
second_analysis_result = {
    'insight_type': 'percentage_drivers_with_second_language',
    'value': round(second_lang_percentage, 2)
}

all_ride_data = pd.DataFrame()
for file_num in range(1, 5):
    ride_file_path = rf'D:\chentan\RNN_Numpy\rides_{file_num}.csv'
    ride_data = pd.read_csv(ride_file_path)
    all_ride_data = pd.concat([all_ride_data, ride_data], ignore_index=True)

successful_rides = all_ride_data[all_ride_data['status'] == 'Success']
ride_success_percentage = (len(successful_rides) / len(all_ride_data)) * 100
third_analysis_result = {
    'insight_type': 'ride_success_rate',
    'value': round(ride_success_percentage, 2)
}

all_results = pd.DataFrame([first_analysis_result, second_analysis_result, third_analysis_result])

output_file = 'analysis_results.csv'
all_results.to_csv(output_file, index=False)
```

```python
# 深度学习训练相关名词解释

# 1.简述 Batch Normalization 的基本原理与作用。
# Batch Normalization 是一种常用的归一化 通常适用于神经网络的每一层输出
# 作用：加速收敛：通过归一化输入，使得每一层的输入分布更加稳定，减少了内部协变量偏移（Internal Covariate Shift）的问题，从而加快了模型的收敛速度
# 。提高泛化能力：可以在一定程度上起到正则化的作用，减少过拟合的风险。例如，在图像分类任务中，使用 Batch Normalization 可以让模型更快地收敛到较好的性能


# 2.什么是 Dropout？它在深度神经网络训练中的作用是什么？
# Dropout 是一种正则化技术，在训练过程中，以一定的概率P随机丢弃网络中的一些神经元
# 作用：防止过拟合：通过随机丢弃神经元，使得模型不能过度依赖某些特定的神经元，从而迫使模型学习到更鲁棒的特征
# 模型集成的近似：每次丢弃不同的神经元组合，相当于训练了多个不同的子模型，在预测时将这些子模型的结果进行平均，类似于集成学习的效果


# 3.解释梯度裁剪的概念及其在训练过程中的作用。
#梯度裁剪是指在训练过程中，对梯度的大小进行限制。当梯度的范数（通常是 L2 范数）超过某个预设的阈值时，将梯度进行缩放，使其范数不超过该阈值。
# 作用： 防止梯度爆炸：在深度神经网络中，特别是在使用循环神经网络（RNN）时，梯度可能会在反向传播过程中变得非常大，导致模型无法收敛。梯度裁剪可以有效地限制梯度的大小，避免梯度爆炸的问题。
# 稳定训练过程：通过控制梯度的范围，使得训练过程更加稳定，减少了因梯度过大而导致的参数更新不稳定的情况


# 4.什么是早停？为什么它能防止过拟合？
# 早停是一种在训练过程中防止过拟合的策略。在训练过程中，将数据集分为训练集和验证集，在每次训练迭代后，在验证集上评估模型的性能。当验证集上的性能不再提升，停止训练。
# 防止过拟合的原因
# 随着训练的进行，模型在训练集上的性能会不断提高，但可能会开始对训练数据过拟合，导致在验证集上的性能下降。早停通过在验证集性能达到最优时停止训练，避免了模型在训练集上太过拟合，提高了模型泛化能力


# 5.数据增强（Data Augmentation）在训练过程中的意义是什么？请举例说明。
#意义
# 数据增强是指在不增加实际数据的情况下，通过对原始数据进行各种变换来生成新的数据样本其意义在于：
# 增加数据多样性：可以扩充训练数据的规模，使模型能够学习到更多不同的特征，提高模型的泛化能力。
# 减少过拟合：让模型在更多样化的数据上进行训练，降低对训练数据中特定模式的依赖。
#示例：
# 图像数据：在图像分类任务中，可以对图像进行旋转、翻转、缩放、裁剪、添加噪声等操作。例如，将一张猫的图像进行水平翻转，得到一张新的图像，这两张图像虽然内容相似，但在像素层面有所不同，可以作为不同的样本用于训练。
# 文本数据：可以对文本进行同义词替换、插入或删除一些单词等操作。例如，将句子“这是一只可爱的猫”中的“可爱”替换为“乖巧”，得到一个新的句子用于训练。


# 6.什么是梯度消失？ 为什么会出现？ 有哪些解决办法？
# 梯度消失是指在深度神经网络反向传播中，随着网络层数越来越多梯度会越来越小，使得靠近输入层的神经元的参数更新非常缓慢，甚至几乎不更新。
# 原因
# 激活函数的选择：一些激活函数（如 Sigmoid 函数和 Tanh 函数）的导数在输入值较大或较小时趋近于 0。当网络层数较多时，经过多次求导后，梯度会变得非常小。
# 网络结构过深：随着网络层数的增加，梯度在反向传播过程中不断被乘以权重矩阵，导致梯度逐渐消失。
# 解决办法
# 使用合适的激活函数：如 ReLU（Rectified Linear Unit）函数，其导数在正数区间恒为 1，避免了梯度消失的问题。
# Batch Normalization：可以使得每一层的输入分布更加稳定，减少了梯度消失的影响。
# 残差连接：在深度残差网络（ResNet）中，通过残差块的设计，允许梯度直接跳过一些层进行传播，缓解了梯度消失的问题


# 7.什么是学习率调度？有什么好处？
#学习率调度是指在训练过程中，根据训练的进展动态地调整学习率的大小。常见的学习率调度策略有学习率衰减（如按固定步长衰减、指数衰减等）。
# 好处
# 加速收敛：在训练初期，使用较大的学习率可以让模型快速地接近最优解。
# 避免错过最优解：在训练后期，随着模型逐渐接近最优解，减小学习率可以避免模型因为学习率过大而跳过最优解，从而提高模型的性能。例如，在训练一个复杂的图像生成模型时，使用学习率调度可以让模型在不同阶段都能更有效地进行参数更新。
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dropout, Dense

# 假设我们用 60 天的价格预测第 61 天价格
time_steps = 60
features = 1
samples = 200

# 构造输入数据 (samples, time_steps, features)
x_train = np.random.rand(samples, time_steps, features)  # 200个 60天股票价格
y_train = np.random.rand(samples, 1)                     # 200个 对应的第61天股票价格
x_test = np.random.rand(10, time_steps, features)        # 测试数据

# 1. 使用 stock_model 变量构建模型
tkstock = Sequential()
tkstock.add(LSTM(units=4, input_shape=(time_steps, features), return_sequences=True))
tkstock.add(Dropout(rate=0.2))
tkstock.add(LSTM(units=4))
tkstock.add(Dropout(rate=0.2))
tkstock.add(Dense(units=1))

# 2. 编译模型
tkstock.compile(loss='mean_squared_error', optimizer='adam')

# 3. 训练配置
tkstock.fit(x_train, y_train, epochs=5, batch_size=16, verbose=0)

# 4. 输出模型 summary()
print("模型结构信息：")
tkstock.summary()

# 5. 预测 x_test 并输出最后一列
predictions = tkstock.predict(x_test)
print("对 x_test 的预测结果最后一列：", predictions.flatten())
```
